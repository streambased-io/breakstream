```python
import numpy as np
from datetime import datetime, timezone
from matplotlib.lines import Line2D
from matplotlib.patches import Ellipse

def plot_frac_job(job, telemetry_df, stages_df):
    """Wellbore schematic with every stage marked, above the job's treatment chart."""
    well_name = job['well_name'] if isinstance(job['well_name'], str) else job['wellId'][:8]
    completed = stages_df[stages_df['state'] == 'completed']
    screenouts = stages_df[stages_df['state'] == 'screenout']
    job_time = datetime.fromtimestamp(job['end_ts'] / 1000, tz=timezone.utc).strftime('%Y-%m-%d %H:%M:%S UTC')

    fig, (ax_well, ax_tc) = plt.subplots(2, 1, figsize=(14, 10), gridspec_kw={'height_ratios': [1, 1.1]})

    # ── Wellbore: vertical section, build curve, then the lateral (x = horizontal displacement)
    R = 900                                              # build radius, ft
    heel_md = stages_df['measured_depth_ft'].min() - 400
    toe_md = stages_df['measured_depth_ft'].max() + 400
    kop = heel_md - np.pi / 2 * R                        # kick-off point (TVD == MD above it)
    landing_tvd = kop + R
    lateral = lambda md: (R + (md - heel_md), landing_tvd + 25 * np.sin((md - heel_md) / 900))

    curve = np.linspace(0, np.pi / 2, 60)
    lat_md = np.linspace(heel_md, toe_md, 200)
    xs = np.concatenate([[0, 0], R * (1 - np.cos(curve)), lateral(lat_md)[0]])
    ys = np.concatenate([[0, kop], kop + R * np.sin(curve), lateral(lat_md)[1]])

    # Strata, with the target shale around the lateral
    x_max = R + (toe_md - heel_md) + 600
    bands = [(0, kop * 0.35, '#e8dcc0'), (kop * 0.35, kop * 0.7, '#d9c9a3'),
             (kop * 0.7, landing_tvd - 250, '#c8b48a'), (landing_tvd - 250, landing_tvd + 250, '#5d5048'),
             (landing_tvd + 250, landing_tvd + 1600, '#a89878')]
    for top, bottom, colour in bands:
        ax_well.axhspan(top, bottom, color=colour, zorder=0)
    ax_well.text(x_max - 80, landing_tvd + 1450, 'fracture height exaggerated', color='#555', ha='right', fontsize=8, zorder=1)
    ax_well.text(x_max - 80, landing_tvd - 160, 'target shale', color='#eee', ha='right', fontsize=9, zorder=1)

    ax_well.plot(xs, ys, color='#222', linewidth=3, zorder=3, solid_capstyle='round')
    ax_well.scatter([0], [0], marker='^', s=260, color='#37474f', zorder=5)
    ax_well.text(80, -120, 'wellhead', fontsize=9, color='#37474f')

    for _, s in stages_df.iterrows():
        x, y = lateral(s['measured_depth_ft'])
        ok = s['state'] == 'completed'
        half_height = 350 + 900 * s['proppant_placed_lbs'] / 450_000   # fracture height ~ proppant placed (exaggerated)
        ax_well.add_patch(Ellipse((x, y), 220, 2 * half_height, alpha=0.5, zorder=2,
                                  color='#ffb300' if ok else '#e53935'))
        ax_well.scatter([x], [y], s=110, zorder=4, marker='o' if ok else 'X',
                        color='#2ca02c' if ok else '#d62728', edgecolors='white', linewidths=0.8)
        ax_well.text(x, y - half_height - 120, str(s['stage_number']), ha='center', fontsize=8, color='#333')

    ax_well.set_xlim(-600, x_max)
    ax_well.set_ylim(landing_tvd + 1600, -300)
    ax_well.set_xlabel('horizontal displacement (ft)')
    ax_well.set_ylabel('true vertical depth (ft)')
    ax_well.legend(handles=[
        Line2D([], [], marker='o', color='#2ca02c', linestyle='', markersize=9, label=f'completed ({len(completed)})'),
        Line2D([], [], marker='X', color='#d62728', linestyle='', markersize=9, label=f'screenout ({len(screenouts)})'),
    ], loc='upper right')
    ax_well.set_title(f'{well_name}  ·  {len(stages_df)} stages, {len(completed)} completed, '
                      f'{len(screenouts)} screened out  ·  job completed {job_time}')

    # ── Treatment chart
    t0 = telemetry_df['timestamp'].min()
    hours = (telemetry_df['timestamp'] - t0) / 3_600_000
    ax_rate, ax_prop = ax_tc.twinx(), ax_tc.twinx()
    ax_prop.spines['right'].set_position(('axes', 1.07))

    ax_prop.fill_between(hours, telemetry_df['proppant_conc_ppa'], color='#ffb300', alpha=0.35, step='mid')
    ax_rate.plot(hours, telemetry_df['slurry_rate_bpm'], color='#1e88e5', linewidth=1.2)
    ax_tc.plot(hours, telemetry_df['treating_pressure_psi'], color='#c62828', linewidth=1.4)
    for ts in screenouts['timestamp']:
        ax_tc.axvline((ts - t0) / 3_600_000, color='#d62728', linestyle='--', alpha=0.6)

    ax_tc.set_zorder(ax_prop.get_zorder() + 1)
    ax_tc.patch.set_visible(False)
    ax_tc.set_xlabel('hours since job start')
    ax_tc.set_ylabel('treating pressure (psi)', color='#c62828')
    ax_rate.set_ylabel('slurry rate (bbl/min)', color='#1e88e5')
    ax_prop.set_ylabel('proppant (lb/gal)', color='#b28704')
    ax_rate.set_ylim(0, 130)
    ax_prop.set_ylim(0, 8)
    ax_rate.grid(False)
    ax_prop.grid(False)
    ax_tc.legend(handles=[
        Line2D([], [], color='#c62828', label='treating pressure'),
        Line2D([], [], color='#1e88e5', label='slurry rate'),
        Line2D([], [], color='#ffb300', linewidth=6, alpha=0.5, label='proppant concentration'),
        Line2D([], [], color='#d62728', linestyle='--', label='screenout'),
    ], loc='upper left', fontsize=8, ncol=4)
    ax_tc.set_title('Treatment chart')

    plt.tight_layout()
    plt.show()
```

```python
import matplotlib.patheffects as pe
import matplotlib.ticker as mticker
from matplotlib.lines import Line2D
from matplotlib.patches import Patch

def find_breakout(envelope, live_curve):
    """First reading after which the live stage stays above the historical P95 until the screenout."""
    m = live_curve.merge(envelope[['bucket', 'p95']], on='bucket', how='left').sort_values('timestamp')
    above = ((m['net_psi'] > m['p95']) & (m['progress'] > 0.2)).tolist()
    for i in range(len(above)):
        if all(above[i:]):
            return m.iloc[i]
    return None

def plot_pressure_envelope(live, envelope, past_screenouts, live_curve):
    breakout = find_breakout(envelope, live_curve)
    lead_seconds = None if breakout is None else (live['screenout_ts'] - breakout['timestamp']) / 1000
    well = live['well_name'] if isinstance(live['well_name'], str) else live['wellId'][:8]
    n_hist = int(envelope['stages'].max())
    n_past = past_screenouts.groupby(['wellId', 'stage_number']).ngroups
    centre = 1 / 80   # buckets are 1/40 wide; plot them at their centres

    LIVE, BAND, BAND_CORE, MEDIAN, PAST = '#ff6d00', '#90caf9', '#1e88e5', '#0d47a1', '#e53935'

    with plt.style.context('seaborn-v0_8-white'):
        fig, ax = plt.subplots(figsize=(14, 7.5))
        ax.set_facecolor('#fbfcfe')

        # stage phases
        for x0, x1, label in [(0, 0.15, 'PAD'), (0.15, 0.85, 'PROPPANT RAMP'), (0.85, 1.0, 'FLUSH')]:
            ax.text((x0 + x1) / 2, 1.012, label, transform=ax.get_xaxis_transform(), ha='center',
                    va='bottom', fontsize=8.5, color='#90a4ae', fontweight='bold')
        for x in (0.15, 0.85):
            ax.axvline(x, color='#cfd8dc', lw=1, ls=':', zorder=0)

        # every past screenout, faintly
        for _, g in past_screenouts.groupby(['wellId', 'stage_number']):
            g = g.sort_values('bucket')
            ax.plot(g['bucket'] + centre, g['net_psi'], color=PAST, alpha=0.10, lw=1, zorder=1)

        # the historical envelope
        x = envelope['bucket'] + centre
        ax.fill_between(x, envelope['p05'], envelope['p95'], color=BAND, alpha=0.45, lw=0, zorder=2)
        ax.fill_between(x, envelope['p25'], envelope['p75'], color=BAND_CORE, alpha=0.35, lw=0, zorder=2)
        ax.plot(x, envelope['p50'], color=MEDIAN, lw=1.4, ls='--', zorder=3)

        # the live stage
        end = live_curve.iloc[-1]
        if breakout is not None:
            ax.axvspan(breakout['progress'], end['progress'], color=LIVE, alpha=0.07, lw=0, zorder=0)
        ax.plot(live_curve['progress'], live_curve['net_psi'], color=LIVE, lw=3, zorder=5,
                solid_capstyle='round', path_effects=[pe.Stroke(linewidth=7, foreground='white'), pe.Normal()])
        ax.scatter([end['progress']], [end['net_psi']], marker='X', s=220, color=PAST,
                   edgecolors='white', linewidths=1.5, zorder=7, clip_on=False)
        ax.annotate('screenout', (end['progress'], end['net_psi']), xytext=(-14, 0),
                    textcoords='offset points', ha='right', va='center', fontsize=10,
                    fontweight='bold', color=PAST, zorder=7)

        if breakout is not None:
            ax.scatter([breakout['progress']], [breakout['net_psi']], s=180, facecolor='white',
                       edgecolors=LIVE, linewidths=3, zorder=6)
            mins, secs = divmod(int(round(lead_seconds)), 60)
            ax.annotate(f'Leaves the historical envelope\n{mins} min {secs:02d} s before the screenout',
                        (breakout['progress'], breakout['net_psi']), xytext=(-170, 70),
                        textcoords='offset points', fontsize=11, fontweight='bold', color='#bf360c',
                        ha='center', zorder=8,
                        bbox=dict(boxstyle='round,pad=0.6', fc='white', ec=LIVE, lw=1.5),
                        arrowprops=dict(arrowstyle='-|>', color=LIVE, lw=1.8,
                                        connectionstyle='arc3,rad=-0.25'))

        top = max(live_curve['net_psi'].max(), envelope['p95'].max()) * 1.12
        ax.set_ylim(min(envelope['p05'].min(), live_curve['net_psi'].min()) - 250, top)
        ax.set_xlim(0, 1.015)
        ax.xaxis.set_major_formatter(mticker.PercentFormatter(1.0))
        ax.yaxis.set_major_formatter(mticker.FuncFormatter(lambda v, _: f'{v:+,.0f}'))
        ax.set_xlabel('progress through the stage', fontsize=10, color='#455a64')
        ax.set_ylabel('pressure above the stage\'s own baseline (psi)', fontsize=10, color='#455a64')
        ax.grid(axis='y', color='#eceff1', lw=1)
        for side in ('top', 'right'):
            ax.spines[side].set_visible(False)
        for side in ('left', 'bottom'):
            ax.spines[side].set_color('#b0bec5')
        ax.tick_params(colors='#607d8b')

        ax.legend(handles=[
            Line2D([], [], color=LIVE, lw=3, label=f'live: {well}, stage {live["stage_number"]}  (isk.hotset)'),
            Patch(color=BAND_CORE, alpha=0.5, label='typical stage, middle 50%  (isk.union)'),
            Patch(color=BAND, alpha=0.6, label='historical P5–P95 envelope'),
            Line2D([], [], color=MEDIAN, lw=1.4, ls='--', label='historical median'),
            Line2D([], [], color=PAST, alpha=0.5, lw=1, label=f'past screenouts ({n_past:,})'),
        ], loc='upper left', fontsize=9, frameon=True, framealpha=0.95, edgecolor='#e0e0e0')

        fig.suptitle(f'{well} · stage {live["stage_number"]} against {n_hist:,} historical stages',
                     x=0.06, ha='left', fontsize=15, fontweight='bold', color='#1a237e')
        fig.text(0.06, 0.915, 'Treating pressure, aligned by stage progress. The blue envelope is what normal '
                 'has looked like on every completed stage so far.', fontsize=10, color='#607d8b')
        plt.tight_layout(rect=(0, 0, 1, 0.93))
        plt.show()

    return {'well': well, 'stage': int(live['stage_number']), 'historical_stages': n_hist,
            'past_screenouts': n_past, 'lead_seconds': lead_seconds}
```

```python
# 1. Every stage ever pumped - archived and live - as a pressure curve,
#    aligned on how far through the stage each reading was taken
spark.sql("""
CREATE OR REPLACE TEMP VIEW stage_curves AS
WITH readings AS (
  SELECT wellId, stage_number, timestamp, treating_pressure_psi,
         (timestamp - MIN(timestamp) OVER w) / (MAX(timestamp) OVER w - MIN(timestamp) OVER w) AS progress
  FROM isk.union.pump_telemetry
  WINDOW w AS (PARTITION BY wellId, stage_number)
),
baseline AS (  -- each stage's own pressure once the pad is pumping at full rate
  SELECT wellId, stage_number, AVG(treating_pressure_psi) AS baseline_psi
  FROM readings
  WHERE progress BETWEEN 0.08 AND 0.18
  GROUP BY wellId, stage_number
)
SELECT r.wellId, r.stage_number, f.state, r.timestamp, r.progress,
       FLOOR(r.progress * 40) / 40 AS bucket,
       r.treating_pressure_psi - b.baseline_psi AS net_psi
FROM readings r
JOIN baseline b             ON r.wellId = b.wellId AND r.stage_number = b.stage_number
JOIN isk.union.frac_stages f ON r.wellId = f.wellId AND r.stage_number = f.stage_number
""")

# 2. The fleet's most recent screenout - straight from the live hotset
live = q("""
SELECT f.wellId, f.stage_number, f.timestamp AS screenout_ts,
       get_json_object(c.data, '$.well_name') AS well_name
FROM isk.hotset.frac_stages f
LEFT JOIN isk.hotset.well_control_events c
       ON c.wellId = f.wellId AND c.event_type = 'job_summary'
WHERE f.state = 'screenout'
ORDER BY f.timestamp DESC
LIMIT 1
""").iloc[0]

# 3. What "normal" looks like: the spread of every other completed stage
envelope = q(f"""
SELECT bucket,
       percentile_approx(net_psi, 0.05) AS p05,
       percentile_approx(net_psi, 0.25) AS p25,
       percentile_approx(net_psi, 0.50) AS p50,
       percentile_approx(net_psi, 0.75) AS p75,
       percentile_approx(net_psi, 0.95) AS p95,
       COUNT(DISTINCT wellId, stage_number) AS stages
FROM stage_curves
WHERE state = 'completed' AND wellId <> '{live['wellId']}'
GROUP BY bucket
ORDER BY bucket
""")

# 4. What screenouts have looked like before
past_screenouts = q(f"""
SELECT wellId, stage_number, bucket, AVG(net_psi) AS net_psi
FROM stage_curves
WHERE state = 'screenout' AND wellId <> '{live['wellId']}'
GROUP BY wellId, stage_number, bucket
""")

# 5. The live stage itself
live_curve = q(f"""
SELECT timestamp, progress, bucket, net_psi
FROM stage_curves
WHERE wellId = '{live['wellId']}' AND stage_number = {live['stage_number']}
ORDER BY timestamp
""")

warning = plot_pressure_envelope(live, envelope, past_screenouts, live_curve)
```

```python
import pandas as pd
import json
import subprocess
import time
import os
from IPython.display import HTML, display
import matplotlib.pyplot as plt
from pyspark.sql import SparkSession

spark = SparkSession.builder.getOrCreate()
```

```python
def compose(*args, capture=False, timeout=10):
    host_dir = os.environ["BREAKSTREAM_HOST_DIR"]
    cmd = [
        "docker", "compose",
        "--project-directory", f"{host_dir}/environment",
        "-f", f"{host_dir}/environment/docker-compose.yaml",
    ] + list(args)
    print(">", " ".join(cmd))
    try:
        result = subprocess.run(cmd, check=True, capture_output=capture, text=True, timeout=timeout)
    except subprocess.TimeoutExpired as e:
        # Expected for console consumers asked for more messages than the topic
        # actually has (e.g. --max-messages 10 on a shorter topic) — it just sits
        # there waiting for more, so cut it off rather than hang the cell forever.
        print(f"(stopped after {timeout}s — expected if fewer messages were available than requested)")
        return e.stdout if capture else None
    if capture:
        return result.stdout
```

```python
def show_distribution(table='pump_telemetry'):

    def safe_int(v):
        if v is None or pd.isna(v):
            return None
        return int(v)
    
    q(f"REFRESH TABLE isk.hotset.{table}")
    q(f"REFRESH TABLE direct.coldset.{table}")
    hot_offset_peg = safe_int(q(f"SELECT MAX(kafka_offset) as kafka_offset FROM isk.hotset.{table}").iloc[0]['kafka_offset']) or 0
    cold_offset_peg = safe_int(q(f"SELECT MAX(kafka_offset) as kafka_offset FROM isk.coldset.{table}").iloc[0]['kafka_offset']) or 0
    offset_peg = max(hot_offset_peg, cold_offset_peg)
    hot_row = q(f"SELECT COUNT(*) AS cnt, MIN(kafka_offset) AS earliest, MAX(kafka_offset) AS latest FROM isk.hotset.{table} where kafka_offset <= {offset_peg}").iloc[0]
    cold_row = q(f"SELECT COUNT(*) AS cnt, MIN(kafka_offset) AS earliest, MAX(kafka_offset) AS latest FROM direct.coldset.{table} where kafka_offset <= {offset_peg}").iloc[0]
    union_row = q(f"SELECT COUNT(*) AS cnt FROM isk.union.{table} where kafka_offset <= {offset_peg}").iloc[0]

    hot_count = safe_int(hot_row['cnt']) or 0
    cold_count = safe_int(cold_row['cnt']) or 0
    union_count = safe_int(union_row['cnt']) or 0

    hot_earliest = safe_int(hot_row['earliest'])
    hot_latest = safe_int(hot_row['latest'])
    cold_earliest = safe_int(cold_row['earliest'])
    cold_latest = safe_int(cold_row['latest'])

    def fmt_offset(v):
        return f"{v:,}" if v is not None else '—'

    denom = max(union_count, 1)
    hot_pct = round(100 * hot_count / denom)
    cold_pct = round(100 * cold_count / denom)

    total_pct = hot_pct + cold_pct
    if total_pct == 0:
        hot_bar_w = 0
        cold_bar_w = 0
        empty_bar = True
    else:
        hot_bar_w = hot_pct / total_pct * 100
        cold_bar_w = cold_pct / total_pct * 100
        empty_bar = False

    def bar_label(label, width):
        if width > 15:
            return f'<span style="color:#fff;font-size:12px;font-weight:600;pointer-events:none">{label}</span>'
        return ''

    if empty_bar:
        bar_html = '<div style="width:100%;height:32px;background:#e9ecef;border-radius:4px;"></div>'
    else:
        bar_html = f'''
        <div style="display:flex;width:100%;height:32px;border-radius:4px;overflow:hidden;">
          <div style="width:{hot_bar_w:.1f}%;background:#e63946;display:flex;align-items:center;justify-content:center;">
            {bar_label('Hot Set', hot_bar_w)}
          </div>
          <div style="width:{cold_bar_w:.1f}%;background:#4cc9f0;display:flex;align-items:center;justify-content:center;">
            {bar_label('Cold Set', cold_bar_w)}
          </div>
        </div>'''

    def svg_ring(pct, color):
        r = 40
        cx = cy = 50
        circumference = 2 * 3.14159 * r
        dash = circumference * min(pct, 100) / 100
        gap = circumference - dash
        return f'''
        <svg width="100" height="100" viewBox="0 0 100 100">
          <circle cx="{cx}" cy="{cy}" r="{r}" fill="none" stroke="#e9ecef" stroke-width="8"/>
          <circle cx="{cx}" cy="{cy}" r="{r}" fill="none" stroke="{color}" stroke-width="8"
            stroke-dasharray="{dash:.2f} {gap:.2f}"
            stroke-linecap="round"
            transform="rotate(-90 {cx} {cy})"/>
          <text x="{cx}" y="{cy}" text-anchor="middle" dominant-baseline="central"
            style="font-size:16px;font-weight:700;fill:#222;">{pct}%</text>
        </svg>'''

    html = f'''
    <div style="font-family:system-ui,sans-serif;border:1.5px solid #ff6b35;border-radius:8px;padding:16px;max-width:680px;">
      <div style="margin-bottom:8px;">
        <span style="color:#ff6b35;font-weight:700;font-size:15px;">Union Set</span>
        <span style="color:#222;font-size:15px;margin-left:16px;">Total Rows: <strong>{union_count:,}</strong></span>
      </div>
      {bar_html}
      <div style="display:flex;gap:16px;margin-top:16px;">
        <div style="flex:1;border:1.5px solid #e63946;border-radius:6px;padding:12px;">
          <div style="display:flex;justify-content:space-between;align-items:baseline;margin-bottom:8px;">
            <span style="color:#e63946;font-weight:700;font-size:14px;">Hot Set</span>
            <span style="color:#999;font-size:12px;">kafka</span>
          </div>
          <div style="display:flex;justify-content:center;margin:8px 0;">
            {svg_ring(hot_pct, '#e63946')}
          </div>
          <table style="width:100%;font-size:13px;border-collapse:collapse;">
            <tr><td style="color:#555;">Total Messages</td><td style="text-align:right;font-weight:600;">{hot_count:,}</td></tr>
            <tr><td style="color:#555;">Earliest Offset</td><td style="text-align:right;">{fmt_offset(hot_earliest)}</td></tr>
            <tr><td style="color:#555;">Latest Offset</td><td style="text-align:right;">{fmt_offset(hot_latest)}</td></tr>
          </table>
        </div>
        <div style="flex:1;border:1.5px solid #4cc9f0;border-radius:6px;padding:12px;">
          <div style="display:flex;justify-content:space-between;align-items:baseline;margin-bottom:8px;">
            <span style="color:#4cc9f0;font-weight:700;font-size:14px;">Cold Set</span>
            <span style="color:#999;font-size:12px;">iceberg</span>
          </div>
          <div style="display:flex;justify-content:center;margin:8px 0;">
            {svg_ring(cold_pct, '#4cc9f0')}
          </div>
          <table style="width:100%;font-size:13px;border-collapse:collapse;">
            <tr><td style="color:#555;">Total Rows</td><td style="text-align:right;font-weight:600;">{cold_count:,}</td></tr>
            <tr><td style="color:#555;">Earliest Offset</td><td style="text-align:right;">{fmt_offset(cold_earliest)}</td></tr>
            <tr><td style="color:#555;">Latest Offset</td><td style="text-align:right;">{fmt_offset(cold_latest)}</td></tr>
          </table>
        </div>
      </div>
    </div>
    '''
    display(HTML(html))
```

### Force Kafka retention to roll off early

In production, Kafka would only delete old messages after the configured retention window (often days). To see that behaviour without waiting, the two cells below temporarily lower the `pump_telemetry` topic's config directly:

- `segment.ms=500` — closes the active log segment every 500ms instead of waiting for the default schedule, so old segments become eligible for deletion almost immediately.
- `retention.ms=500` — tells Kafka to delete any closed segment older than 500ms.

Together, these force Kafka to delete the topic's oldest messages within seconds. Both settings get restored to their normal values later in this section.

> **Note:** this isn't special Streambased behaviour — it's completely normal Kafka retention. We're just turning the dial up so it happens in seconds instead of days, since a notebook demo can't sit around waiting for a real retention window to pass.


```python
compose('exec', 'kafka1', 'kafka-configs', '--bootstrap-server', 'kafka1:9092', '--alter', '--topic', 'pump_telemetry', '--add-config', 'segment.ms=500')
```

### Confirm retention has kicked in

Rather than checking in an external UI, we can ask `isk.hotset` directly: once Kafka has actually deleted the oldest segments, the earliest visible `kafka_offset` will move past `0`.

This can take up to a minute to show up here even though Kafka has already deleted the data — Streambased caches Kafka offset ranges internally and refreshes that cache roughly every 60 seconds, on top of Spark's own catalog cache. The cell below polls until it sees the change (or times out).


```python
import time

def wait_for_retention(table='pump_telemetry', timeout_s=90, poll_s=5):
    start = time.time()
    while time.time() - start < timeout_s:
        q(f"REFRESH TABLE isk.hotset.{table}")
        q(f"REFRESH TABLE direct.coldset.{table}")
        row = q(f"SELECT MIN(kafka_offset) AS earliest, COUNT(*) AS cnt FROM isk.hotset.{table}").iloc[0]
        earliest, count = row['earliest'], row['cnt']
        elapsed = int(time.time() - start)
        # retention has kicked in once either the earliest visible offset has moved past 0,
        # or — if it deleted everything currently on the topic — the table is just empty
        # (in which case MIN(kafka_offset) is NULL, not > 0, so count==0 must be checked too)
        if count == 0 or (earliest is not None and earliest > 0):
            print(f"Retention has kicked in — earliest visible offset is now {earliest}, {count} row(s) visible (took {elapsed}s)")
            return earliest
        print(f"Still waiting for retention to take effect... earliest offset is {earliest}, {count} row(s) visible ({elapsed}s elapsed)")
        time.sleep(poll_s)
    print("Timed out waiting for retention to take effect — Streambased's offset cache may just need a little longer. Try re-running this cell.")
    return None

wait_for_retention()
```

### The same topic, unwrapped into a live table

Now query the identical Kafka topic through `isk.hotset` instead. No connector was configured
and no materialized view was built — Streambased recognized the envelope shape and did the
unwrapping automatically.

Notice the schema below: it's the flat `WorkOrderValue` shape (`OrderID`, `WellID`, `Equipment`,
`Status`, `EstimatedCost`), not the raw envelope. And the row count: with only 5 work-order keys in
play, there are at most 5 rows — one per key, always showing its latest state.

As with the retention cache in Part 5, Streambased refreshes its view of which topics exist on
a short cycle rather than instantaneously — so if `orders` was only just created, the cell below
waits for it to show up rather than failing outright.


```python
import time

def wait_for_orders_table(timeout_s=90, poll_s=5):
    start = time.time()
    while time.time() - start < timeout_s:
        try:
            spark.sql("REFRESH TABLE isk.hotset.orders")
            count = q("SELECT COUNT(*) AS cnt FROM isk.hotset.orders").iloc[0]['cnt']
            elapsed = int(time.time() - start)
            print(f"isk.hotset.orders is available — {count} row(s) visible (took {elapsed}s)")
            return True
        except Exception:
            elapsed = int(time.time() - start)
            print(f"Still waiting for Streambased to discover the orders topic... ({elapsed}s elapsed)")
            time.sleep(poll_s)
    print("Timed out waiting for isk.hotset.orders to appear — Streambased's topic discovery may just need a little longer. Try re-running this cell.")
    return False

wait_for_orders_table()
```

### Watching rows update in place

The background generator (`live_datagen.py`) cycles through the same 5 work-order keys, emitting one
create, update, or delete every second. Because I.S.K. applies each change in place, querying
`isk.hotset.orders` repeatedly shows that *same* small set of rows mutating live — no growing
history, no duplicates.

The cell below runs `SELECT * FROM isk.hotset.orders` several times, a couple of seconds apart,
and highlights whatever changed since the previous poll: **yellow** for a row that was created or
updated, **red** for a key that was deleted.


```python
import time
from datetime import datetime
from IPython.display import display, HTML

def watch_orders_cdc(polls=5, interval_s=2):
    prev = {}
    for i in range(polls):
        spark.sql("REFRESH TABLE isk.hotset.orders")
        df = q("SELECT * FROM isk.hotset.orders ORDER BY OrderID")
        cols = list(df.columns)
        cur = {row['OrderID']: row for row in df.to_dict('records')}

        rows_html = ''
        for oid in sorted(set(prev) | set(cur)):
            if oid in cur:
                row = cur[oid]
                changed = oid not in prev or prev[oid] != row
                bg = 'background:#fff3cd;' if changed else ''
                cells = ''.join(
                    f'<td style="padding:6px 12px;border-bottom:1px solid #eee;">{row[c]}</td>'
                    for c in cols
                )
            else:
                bg = 'background:#ffebee;'
                cells = (
                    f'<td style="padding:6px 12px;border-bottom:1px solid #eee;">{oid}</td>'
                    f'<td colspan="{max(len(cols) - 1, 1)}" style="padding:6px 12px;'
                    f'border-bottom:1px solid #eee;color:#c62828;font-style:italic;">'
                    f'deleted since last poll</td>'
                )
            rows_html += f'<tr style="{bg}">{cells}</tr>'

        if not rows_html:
            rows_html = f'<tr><td colspan="{len(cols)}" style="padding:6px 12px;color:#999;">no rows</td></tr>'

        header_html = ''.join(
            f'<th style="padding:6px 12px;text-align:left;border-bottom:2px solid #ccc;'
            f'font-size:0.82em;color:#555;">{c}</th>'
            for c in cols
        )

        legend = (
            ' &nbsp;·&nbsp; <span style="color:#e65100;font-weight:700;">yellow = changed</span>'
            ' &nbsp;·&nbsp; <span style="color:#c62828;font-weight:700;">red = deleted</span>'
        ) if i > 0 else ''

        html = f'''
        <div style="font-family:system-ui,sans-serif;max-width:860px;margin:0 0 14px 0;">
          <div style="font-size:0.85em;color:#666;margin-bottom:6px;">
            <strong>Poll {i + 1}/{polls}</strong> · {datetime.now().strftime('%H:%M:%S')}{legend}
          </div>
          <table style="border-collapse:collapse;font-size:0.85em;width:100%;">
            <tr>{header_html}</tr>
            {rows_html}
          </table>
        </div>
        '''
        display(HTML(html))
        prev = cur
        if i < polls - 1:
            time.sleep(interval_s)

watch_orders_cdc()
```

## Glossary

Quick reference for terms used throughout this notebook.

| Term | Meaning |
|------|---------|
| `isk` | Streambased's unified catalog — sitting above the pre-existing Kafka and Iceberg environments. It surfaces the three logical views below (`isk.hotset`, `isk.coldset`, `isk.union`). |
| `direct` | The underlying Iceberg REST catalog itself, This provides direct (and importantly writable) access to the same physical tables as `isk.coldset`. |
| `hotset` (`isk.hotset`) | Live Kafka topics, surfaced as Iceberg-compatible tables with zero lag — as fresh as the last message on the topic. Subject to Kafka's own retention, so it only holds a recent window of offsets. |
| `coldset` (`isk.coldset` / `direct.coldset`) | Historical data archived into Iceberg tables on object storage. Cheap and durable, and covers the full offset history from 0 up to wherever the hotset begins. |
| `union` (`isk.union`) | The unified view — hotset and coldset stitched into a single table with no duplicates, so one query spans live and archived data together. |
| `kafka_partition` | A column giving each record's original Kafka partition. |
| `kafka_offset` | A column on these tables giving each record's position in its original Kafka partition. Used throughout the notebook to show where the coldset ends and the hotset begins. |
| `op` / `before` / `after` | The three fields of a Debezium-style CDC envelope on the `orders` (work orders) topic: `op` is `c` (create), `u` (update), or `d` (delete); `before`/`after` hold the row's prior/new state. `isk.hotset.orders` applies these in place, so it always shows one current-state row per key rather than the raw event stream. |
| `pump_telemetry` | High-volume surface readings from the frac fleet: treating pressure (psi), slurry rate (barrels/min) and proppant concentration (lb/gal), tagged with the stage being pumped. |
| `frac_stages` | One event per stage when it finishes: `completed`, or `screenout` if proppant bridged in the fractures and pressure ran away. Includes the stage's measured depth and the proppant actually placed. |
| `well_control_events` | `job_start`, `job_end` and a `job_summary` JSON blob (well name, pad, stage counts) for each frac job. |
| Pressure envelope | The range treating pressure normally sits in at each point of a stage, built from every completed stage in history (P5–P95 band). A live stage climbing out of it is the early warning sign of a screenout. |
| Screenout | A stage where proppant packs off the fractures or wellbore, so pressure spikes to the safety limit and pumping has to stop. Usually needs a costly coiled-tubing cleanout. The signature is a steep pressure climb while rate falls. |
